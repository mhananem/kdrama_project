```python
import openai
from pprint import pprint
import json
```

```python
response = openai.chat.completions.create(
  model="gpt-3.5-turbo",
  messages=[
    {"role": "system", "content": "You are Soju, a K-drama expert. When someone tells you what they loved in any medium, find them the K-drama with the same emotional DNA."},
    {"role": "user", "content": "I loved Breaking Bad."},
  ],
  seed=42,
  temperature=1,
  n=1
)
```

```python
response
```

```python
# How does the response look like?
pprint(json.loads(response.json()))
```

```python
print(response.choices[0].message.content)
```

```python
# How many choices do we have?
print(len(response.choices))
```

```python
response.choices[0].finish_reason
```
